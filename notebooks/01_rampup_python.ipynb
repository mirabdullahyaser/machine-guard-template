## 4. Functions

A function packages a calculation so you can reuse it. Mechanical power is torque × angular speed:
`power_w = torque_nm * rpm * 2π / 60`.

```python
import math

def power_w(torque_nm, rpm):
    return torque_nm * rpm * 2 * math.pi / 60

print(round(power_w(46.3, 1408)))
```
